```python
from main import generate_response, explain_concept, summarize_text, answer_question
import torch
import pandas as pd 
```

```python
# Device Setup
device = "mps" if torch.backends.mps.is_available() else "cpu"
```

##  TinyLlama Model

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)
# Load model
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float32
).to(device)
print("Model loaded successfully.")
```

##  Qwen Model

```python
model2 = "Qwen/Qwen2.5-0.5B-Instruct"

tokenizer_two = AutoTokenizer.from_pretrained(model2)

model_qwen = AutoModelForCausalLM.from_pretrained(
    model2,
    torch_dtype=torch.float32
).to(device)
print("Model loaded successfully.")
```

### Define all inputs

```python
text = """
Artificial intelligence is the ability of computer systems to perform tasks that usually require human intelligence.
These tasks include learning, reasoning, problem-solving, understanding language, and recognizing images.
AI is used in many applications such as chatbots, recommendation systems, medical diagnosis, and self-driving cars.
"""
```

```python
prompt ="Explain machine learning in simple words."
```

```python
concept = "Artificial Intelligence"
```

```python
Context= "Transformers use attention mechanisms to process sequences."
Question= "What mechanism do Transformers use?"
```

##  Test Generation

```python
print("Qwen response:\n",generate_response(prompt,150,tokenizer_two,model_qwen))
```

```python
print("TinyLlama response:\n",generate_response(prompt,150,tokenizer,model))
```

##  Explanation Task

```python
model1_explain = explain_concept(concept, tokenizer,model)
```

```python
model2_explain = explain_concept(concept, tokenizer_two,model_qwen)
```

```python
print("tinyllama response:")
print(model1_explain)

print("\n" + "="*40 + "\n")

print("Qwen response:")
print(model2_explain)
```

##  Summarization Task

```python
model1_summary= summarize_text(text, tokenizer,model)
```

```python
model2_summary= summarize_text(text, tokenizer_two,model_qwen)
```

```python
print("tinyllama response:")
print(model1_summary)

print("\n" + "="*40 + "\n")

print("Qwen response:")
print(model2_summary)
```

##  Question Answering Task

```python
model1_answer= answer_question(Context, Question, tokenizer,model)
```

```python
model2_answer = answer_question(Context, Question, tokenizer_two,model_qwen)
```

```python
print("tinyllama response:")
print(model1_answer)

print("\n" + "="*40 + "\n")

print("Qwen response:")
print(model2_answer)
```

##  Comparison

| Prompt | TinyLlama | Qwen | Better | Reason |
|--------|----------|------|--------|--------|
| Explain Concept | Off-topic (hallucination) | Relevant explanation | Qwen | More accurate and focused |
| Summarize | Not a real summary | Structured summary | Qwen | Clear and correct |
| QA | Incorrect (LSTM) | Correct (attention) | Qwen | Uses context correctly |

**1. Which model performed better?**  
Qwen performed better overall.

**2. Which prompt produced the best answer?**  
The question answering prompt.

**3. Did the model hallucinate?**  
Yes, TinyLlama hallucinated. Qwen had minimal hallucination.

**4. What are the limitations of running LLMs locally?**  
Limited resources, slower performance, and lower accuracy with small models.

**5. How can prompt engineering improve the results?**  
By making prompts clear, specific, and structured.
